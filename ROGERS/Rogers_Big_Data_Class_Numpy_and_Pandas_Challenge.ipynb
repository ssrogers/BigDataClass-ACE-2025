```python
# ================================================================
# STEP 3: START YOUR INVESTIGATION
# ================================================================

# Create a copy to work with (always keep original!)
df_clean = df.copy()

print("\n🧹 CLEANING PROCESS BEGINS!")
print("="*70)
```

```python
# ----------------------------------------------------------------
# BUG CATEGORY 1: DUPLICATE RECORDS
# ----------------------------------------------------------------
print("\n1️⃣ Checking for duplicates...")

# TODO: Check for duplicate Patient_IDs
# HINT: Use df_clean['Patient_ID'].duplicated()
#print the rows with repeating data
print(df_clean[df_clean['Patient_ID'].duplicated()])

#correct Patient_ids
df_clean["Patient_ID"]= range(1001, 1001+len(df_clean))
df_clean

#ckeck to see if you still have duplicates in Patient IDs
print(df_clean[df_clean['Patient_ID'].duplicated()])

#check for duplicate names

df_clean.duplicated(subset=['First_Name', 'Last_Name'], keep = False).value_counts()
df_clean[df_clean.duplicated(subset=['First_Name', 'Last_Name'], keep = False)].sort_values(['First_Name','Last_Name'])
# at this level data presents duplicate entries in names but different DOB
#we are going to check which entries actually have similar/different DOB and group them
#any name with a UniqueDOB > 1 means occurs with multiple date of births
DOB_check=df_clean.groupby(['First_Name', 'Last_Name'])['Date_of_Birth'].nunique().reset_index(name='UniqueDOB')
print(DOB_check)

# TODO: Remove or handle duplicates
# HINT: Use drop_duplicates() or investigate why duplicates exist


#print("✓ Duplicate check complete")
```

```python
# ----------------------------------------------------------------
# BUG CATEGORY 2: MISSING VALUES - STANDARDIZE
# ----------------------------------------------------------------
print("\n2️⃣ Standardizing missing values...")

# TODO: Replace all variations of missing values with proper NaN
# HINT: Look for 'NA', 'N/A', empty strings '', and convert to np.nan
# HINT: Use replace() method
#replacing all posible instances of missing data
df_clean.replace(['', ' ', 'NA', 'N/A', 'n/a', 'null', 'NULL', '-'],pd.NA, inplace=True)
df_clean



#print("✓ Missing values standardized")
```

```python
# ----------------------------------------------------------------
# BUG CATEGORY 3: WHITESPACE ISSUES
# ----------------------------------------------------------------
print("\n3️⃣ Cleaning whitespace...")

# TODO: Remove leading/trailing whitespace from text columns
# HINT: Use .str.strip() on string columns
#cleaning can be done per column or for the entire dataset
#per colum route example
    #df_clean['First_Name'] = df_clean['First_Name'].str.strip()
#taking the entire dataset route
df_clean = df_clean.apply(lambda x: x.str.strip() if x.dtype == 'object' else x)
print(df_clean)
# also the column names
df_clean.columns = df_clean.columns.str.strip()
print("✓ Whitespace cleaned")
```

```python
# ----------------------------------------------------------------
# BUG CATEGORY 4: CASE CONSISTENCY
# ----------------------------------------------------------------
print("\n4️⃣ Fixing case inconsistencies...")

# TODO: Standardize capitalization in First_Name, Last_Name, Gender, etc.
# HINT: Use .str.title(), .str.upper(), or .str.lower() as appropriate
# for individual column path
print(df_clean['First_Name'].dtype) #check data type
df_clean['First_Name']= df_clean['First_Name'].apply(lambda x: str(x).strip().title() if pd.notna(x) else '') #use function because we have na values 
print(df_clean['First_Name'].unique())
#found a blank '' in there so have to convert it into NA
df_clean['First_Name'].replace('', pd.NA, inplace=True)
print(df_clean['First_Name'].isna().sum()) # check how many NA have been converted
df_clean['First_Name'].fillna('Benjamin', inplace=True) #change them to 'Benjamin'
print(df_clean['First_Name'].unique()) #check if it has worked
print(df_clean['First_Name'].isna().sum()) #confirm with the sum of Na should return 0

#for Last name colum
df_clean['Last_Name']= df_clean['Last_Name'].apply(lambda x: str(x).strip().title() if pd.notna(x) else '')
print(df_clean['Last_Name'].unique())
#found a blank '' in there so have to convert it into NA
df_clean['Last_Name'].replace('', pd.NA, inplace=True)
print(df_clean['Last_Name'].isna().sum()) # check how many NA have been converted
df_clean['Last_Name'].fillna('Rogers', inplace=True) #change them to 'Rogers'
print(df_clean['Last_Name'].unique()) #check if it has worked
print(df_clean['Last_Name'].isna().sum()) #confirm with the sum of Na should return 0
#for gender column, we need to standardise aswell
print(df_clean['Gender'].dtype)
df_clean['Gender']= df_clean['Gender'].apply(lambda x: str(x).strip().lower() if pd.notna(x) else '')
df_clean['Gender']=df_clean['Gender'].apply(lambda x: 'Male' if x.startswith('m') else 'Female' if x.startswith('f') else 'Unknown') # for any un listed gender, we replace it with unknown
print(df_clean['Gender'].unique())

print("✓ Case consistency applied")
```

```python
# ----------------------------------------------------------------
# BUG CATEGORY 5: DATE FORMAT ISSUES
# ----------------------------------------------------------------
print("\n5️⃣ Converting date columns...")

# TODO: Convert Date_of_Birth, Admission_Date, Discharge_Date to proper datetime
# HINT: Use pd.to_datetime() with errors='coerce'
#coverting DOB first
df_clean['Date_of_Birth'] = df_clean['Date_of_Birth'].astype(str).str.strip().str.replace('/','-', regex=True)   #the astype() converts the column into a string style then strip removes extra whitespaces

df_clean['Date_of_Birth']= pd.to_datetime(df_clean['Date_of_Birth'], errors='coerce') # pd.to_datetime() converts to date/time format
df_clean['Date_of_Birth'].fillna(df_clean['Date_of_Birth'].median().normalize(), inplace=True) #generate values for all NAs (my choice is median). the normarlize helps to remove the time after calculating the median date
print(df_clean['Date_of_Birth'])
#converting ADMISSION DATE 
#print(df_clean['Admission_Date'].head(20))
#print(df_clean['Admission_Date'].unique())
df_clean['Admission_Date'] = df_clean['Admission_Date'].astype(str).str.strip().str.replace('/', '-', regex=True) #standardize the date format(this gave me headache finding the actual cause of NaT)
df_clean['Admission_Date']= pd.to_datetime(df_clean['Admission_Date'], errors='coerce', infer_datetime_format=True)
df_clean['Admission_Date'].fillna(df_clean['Admission_Date'].median().normalize(), inplace=True)
print(df_clean['Admission_Date'])
#coverting DISCHARGE DATE
df_clean['Discharge_Date'] = df_clean['Discharge_Date'].astype(str).str.strip()
df_clean['Discharge_Date']= pd.to_datetime(df_clean['Discharge_Date'], errors='coerce')
df_clean['Discharge_Date'].fillna(df_clean['Discharge_Date'].median().normalize(), inplace=True)
print(df_clean['Discharge_Date'])
print("✓ Dates converted")
```

```python
# ----------------------------------------------------------------
# BUG CATEGORY 6: DATA TYPE CONVERSIONS
# ----------------------------------------------------------------
print("\n6️⃣ Fixing data types...")

# TODO: Convert Age, Heart_Rate to integers (handle strings first)
# HINT: First convert strings to numeric, then to int
#coverting colums to numeric
cols =['Age', 'Heart_Rate']
df_clean[cols]= df_clean[cols].apply(pd.to_numeric, errors='coerce')
#converting to intergers
df_clean[cols] = df_clean[cols].astype('Int64')


# TODO: Convert Bill_Amount to float (remove $ symbols first)
# HINT: Use .str.replace() to remove $ then convert to float
df_clean['Bill_Amount']= df_clean['Bill_Amount'].astype(str).str.replace('$', '')
#coverting to float
df_clean['Bill_Amount'] = df_clean['Bill_Amount'].astype(float)
# generating values for NAs
df_clean['Bill_Amount'].fillna(df_clean['Bill_Amount'].mode(), inplace=True)
print(df_clean['Bill_Amount'].isna().sum) #check if it has worked, should return 0
print(df_clean['Bill_Amount'].unique())

print("✓ Data types corrected")
df_clean
```

```python
# ----------------------------------------------------------------
# BUG CATEGORY 7: INVALID VALUES
# ----------------------------------------------------------------
print("\n7️⃣ Handling invalid values...")

# TODO: Fix negative ages, impossible ages (>120), negative bill amounts
# HINT: Use conditional replacement or set to NaN
#setting  invalids to NaN
df_clean.loc[(df_clean['Age'] < 0)| (df_clean['Age'] > 120), 'Age'] = pd.NA
#filling all nA vales with the median age(or can be any value you want)
df_clean['Age'] = df["Age"].fillna(df['Age'].median())


# TODO: Fix heart rates outside normal range (30-200)
#setting abnormal values to NA
df_clean.loc[(df_clean['Heart_Rate'] < 30)| (df_clean['Heart_Rate'] > 200), 'Heart_Rate'] = pd.NA
#filling all nA vales with the median Heart_Rate(or can be any value you want)
df_clean['Heart_Rate'] = df["Heart_Rate"].fillna(df['Heart_Rate'].median())

# TODO: Fix impossible temperatures (>42°C or <30°C)
#converting temperature column to string first
df_clean['Temperature'].replace('', np.nan, inplace = True) #replace blanks first
df_clean['Temperature']= df_clean['Temperature'].astype(str).str.strip() # also remove extra spaces
#create a mask for farenheit values and 'normal'
mask_F = df_clean['Temperature'].str.endswith(('F', 'f'))
mask_N = df_clean['Temperature'].str.lower()=='normal'
#convert these to celsius
df_clean.loc[mask_F, 'Temperature'] = ((df_clean.loc[mask_F, 'Temperature'].str.rstrip('Ff').astype(float)-32)*5/9) 
df_clean.loc[mask_N, 'Temperature'] = 37.0 

#now convert to float and round off to 1 decimal place
df_clean.loc[mask_F, 'Temperature']= pd.to_numeric(df_clean.loc[mask_F, 'Temperature'], errors='coerce').round(1)
df_clean.loc[mask_N, 'Temperature']= pd.to_numeric(df_clean.loc[mask_N, 'Temperature'], errors='coerce')
df_clean['Temperature']= pd.to_numeric(df_clean['Temperature'],errors='coerce')
                    
#setting imposible values to NA
df_clean.loc[(df_clean['Temperature'] < 30)| (df_clean['Temperature'] > 42), 'Temperature'] = np.nan
#filling all nA vales with the median temperature(or can be any value you want)
print(df_clean['Temperature'].isna().sum)
df_clean['Temperature'].fillna(df_clean['Temperature'].median(), inplace=True)
# check to see if values have changed
print(df_clean.loc[0:50, 'Temperature'])
print(df_clean.loc[50:101, 'Temperature'])

print("✓ Invalid values handled")
df_clean
```

```python
# ----------------------------------------------------------------
# BUG CATEGORY 8: BLOOD TYPE ISSUES
# ----------------------------------------------------------------
print("\n8️⃣ Cleaning Blood Types...")

# TODO: Fix '0+' to 'O+', remove extra spaces, handle invalid blood types
# HINT: Use .replace() or .str.replace()
#find discrepancy in blood groups
print(df_clean.loc[:, 'Blood_Type']) # for viewing the whole dataset but only the column of bloodtype
df_clean['Blood_Type'].unique() # for finding the unique values in the column of dataset
#replace / make similar the values with discrepancies
df_clean['Blood_Type'].replace({'A +': 'A+', '0+':'O+', 'XY+':'AB+'}, inplace=True)
df_clean['Blood_Type'].unique() #check to see if it has worked
#now handle the blanks
print(df_clean['Blood_Type'].isna().sum()) #check and confirm how many blanks you have
df_clean['Blood_Type'].fillna(df_clean['Blood_Type'].mode()[0], inplace=True)
df_clean['Blood_Type'].unique() #check to see if it has worked
print("✓ Blood types cleaned")
df_clean
```

```python
# ----------------------------------------------------------------
# BUG CATEGORY 9: TEMPERATURE UNITS
# ----------------------------------------------------------------
print("\n9️⃣ Standardizing temperature units...")

# TODO: Convert all temperatures to Celsius (remove 'F', convert if needed)
# HINT: Check for 'F' in string, use formula: (F - 32) * 5/9


#done in previous sections

print("✓ Temperatures standardized")
```

```python
# ----------------------------------------------------------------
# BUG CATEGORY 10: BLOOD PRESSURE FORMAT
# ----------------------------------------------------------------
print("\n🔟 Standardizing blood pressure format...")

# TODO: Make all blood pressure values use '/' separator
# HINT: Replace ' / ' and '-' with '/'
#start with finding out the discrepancies
print(df_clean['Blood_Pressure'].unique())
df_clean['Blood_Pressure']= df_clean['Blood_Pressure'].astype(str).str.strip().str.lower()
#now change to '/' for all and also change normal to the most common value
df_clean['Blood_Pressure'].replace({' / ': '/', '-':'/', 'normal': '120/80'}, regex=True, inplace=True)
print(df_clean['Blood_Pressure'].unique()) #check to see if it has worked
print(df_clean.loc[0: 50, 'Blood_Pressure'])
#generate values for NAs
df_clean['Blood_Pressure'].fillna(df_clean['Blood_Pressure'].mode()[0], inplace=True)
print(df_clean['Blood_Pressure'].isna().sum()) # check to see if it has worked
print("✓ Blood pressure standardized")
print(df_clean.loc[0:50, 'Blood_Pressure'])
df_clean
```

```python
# ----------------------------------------------------------------
# BUG CATEGORY 11: INSURANCE PROVIDER CONSISTENCY
# ----------------------------------------------------------------
print("\n1️⃣1️⃣ Standardizing insurance providers...")

# TODO: Make all provider names consistent (e.g., all 'BlueCross' or all 'Aetna')
# HINT: Use .str.lower() then .str.title() or create a mapping
#check items in insurance column
print(df_clean['Insurance_Provider'].unique())
#create a map basing on the unique values
map_IP={
    'CIGNA': 'Cigna',
    'BLUECROSS': 'BlueCross',
    'blue cross': 'BlueCross',
    'UnitedHealth': 'United Health',
    'aetna': 'Aetna'

}
#Use map to standardise
df_clean['Insurance_Provider']= df_clean['Insurance_Provider'].map(map_IP).fillna(df_clean['Insurance_Provider']) #the fillna allows the map to skip values that arent in the map dictionary
print(df_clean['Insurance_Provider'].unique())
#work on the NA values
df_clean['Insurance_Provider'] = df_clean['Insurance_Provider'].fillna(df_clean['Insurance_Provider'].mode()[0]) 
print(df_clean.loc[0:50, 'Insurance_Provider'])  #confirm that it has worked
print("✓ Insurance providers standardized")
```

```python
# ----------------------------------------------------------------
# BUG CATEGORY 12: DIAGNOSIS CLEANUP
# ----------------------------------------------------------------
print("\n1️⃣2️⃣ Cleaning diagnosis field...")

# TODO: Strip whitespace, fix typos (e.g., 'Diabetis' -> 'Diabetes')
# HINT: Use .str.strip() and .replace()
#start with removing white spaces
df_clean['Diagnosis'] = df_clean['Diagnosis'].str.title().str.strip()
#check for spellings
df_clean['Diagnosis'].unique()
df_clean['Diagnosis'].replace({'diabetis': 'Diabetes'}, inplace=True) #replace any that is incorrect
print(df_clean.loc[0:50, 'Diagnosis']) #check to see if has worked

#print("✓ Diagnosis cleaned")
```

```python
# ----------------------------------------------------------------
# BUG CATEGORY 13: LOGICAL INCONSISTENCIES
# ----------------------------------------------------------------
print("\n1️⃣3️⃣ Checking logical consistency...")

# TODO: Ensure Discharge_Date is after Admission_Date
# HINT: Compare the two date columns
# check if Admission date is before discarge date
print(df_clean[df_clean['Admission_Date'] > df_clean['Discharge_Date']])
#for any discharge before admission, we swap the dates
date_mask = df_clean['Admission_Date']> df_clean['Discharge_Date']
df_clean.loc[date_mask, ['Admission_Date', 'Discharge_Date']]= df_clean.loc[date_mask, ['Discharge_Date', 'Admission_Date']].values
print(df_clean[df_clean['Admission_Date'] > df_clean['Discharge_Date']]) # confirm if it has worked, it should return an empty value


# TODO: Calculate proper age from Date_of_Birth
# age will be calculated at the time of admission
df_clean.loc[df_clean['Date_of_Birth'].notna() & df_clean['Admission_Date'].notna(),'Age'] = (df_clean['Admission_Date']- df_clean['Date_of_Birth']).dt.days //365

#print("✓ Logical consistency checked")
df_clean
```
